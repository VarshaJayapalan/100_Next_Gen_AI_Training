Seperate the dataset

```python
X = data.drop("Outcome", axis=1)
Y = data["Outcome"]
```

```python
print("Features shape", X.shape)
```

```python
print("Target shape", Y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

Train test split

```python
X_train, X_test, Y_train,  Y_test = train_test_split(X, Y, test_size=0.85)
```

```python
# Train-Test Split Sizes
print("\nTraining and Testing Dataset Sizes:")
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("Y_train:", Y_train.shape)
print("Y_test :", Y_test.shape)
```

```python
# Find columns containing zero values
zero_columns = (data == 0).sum()
```

```python
print("Columns with zero values:")
print(zero_columns[zero_columns > 0])
```

```python
# Replace zeros with median
for col in data.select_dtypes(include='number').columns:
    median = data[col].median()
    data[col] = data[col].replace(0, median)
```

```python
zero_columns = (data == 0).sum()
```

```python
print("Remaining zero values:")
print(zero_columns[zero_columns > 0])
```

```python
print("\nPREPROCESSING SUMMARY")
print("=" * 40)
print("\nMissing values handled in:")
print(['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI'])

```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import confusion_matrix, classification_report
```
